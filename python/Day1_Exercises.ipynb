# VARIABLE

```python
# Integer Variable
age = 25
print(age)
```

```python
# String Variable
name = "Alice"
print(name)
```

```python
# Float Variable
price = 19.99
print(price)
```

```python
# Boolean Variable
is_active = True
print(is_active)
```

```python
# storing and printing a Value
# Storing the value 10 in a variable called 'x'
x = 10
# Printing the value of 'x'
print(x)
```

```python
# using Variable in Expressions
#Assigning value to variables
a = 5
b = 3
# Adding two variable and storing the result in 'result'
result = a + b
# Printing the result
print(result) 
```

```python
# Changing the value of a variable 
# Initial value
score = 50
print(score)
```

```python
#Changing the value of 'score'
score = 100
print(score)
```

```python
#Concatenating Strings
# Assigning Value to variables
first_name = "John"
last_name = "Doe"

# Concatenating string and storing in a new variable
full_name = first_name+" "+ last_name
print(full_name)
```

```python
#using Variables in a Calculation
#Assigning values to variables
length = 10
width = 5

#Calculatibg the area of the rectangle 
area = length * width
print(area)
```

```python
#Reassigning Values to Variables
x = 10 # Intitial value of x
print(x)
```

```python
x = 20 # Intitial value of x
print(x)
```

```python
first_name = 'Venkat'
last_name = 'Ram'
country = 'IND'
city = 'HYD'
age = 35
is_married = True
skills = ['Autocad', 'BIM', 'Revit', 'python']
person_info = {
    'firstname':'Hanshit', 
    'lastname':'kas', 
    'country':'india',
    'city':'Hyd'
    }



print('First name:', first_name)
print('First name length:', len(first_name))
print('Last name: ', last_name)
print('Last name length: ', len(last_name))
print('Country: ', country)
print('City: ', city)
print('Age: ', age)
print('Married: ', is_married)
print('Skills: ', skills)
print('Person information: ', person_info)
```

```python
# Declaring multiple variables in one line

first_name, last_name, country, age, is_married = 'venkata', 'ram', 'india', 35, True

print(first_name, last_name, country, age, is_married)
print('First name:', first_name)
print('Last name: ', last_name)
print('Country: ', country)
print('Age: ', age)
print('Married: ', is_married)
```
